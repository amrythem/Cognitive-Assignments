```python
L = [10, 20, 30, 40, 50, 60, 70, 80]

# i. Add 200 and 300 to L
L.append(200)
L.append(300)
print("After adding 200 and 300:", L)

# ii. Remove 10 and 30 from L
L.remove(10)
L.remove(30)
print("After removing 10 and 30:", L)

# iii. Sort L in ascending order
L.sort()
print("Ascending order:", L)

# iv. Sort L in descending order
L.sort(reverse=True)
print("Descending order:", L)
```

```python
scores = (45, 89.5, 76, 45.4, 89, 92, 58, 45)

# i. Identify the highest score and its index
highest = max(scores)
print("Highest score:", highest)
print("Index of highest score:", scores.index(highest))

# ii. Find the lowest score and count its occurrences
lowest = min(scores)
print("Lowest score:", lowest)
print("Count of lowest score:", scores.count(lowest))

# iii. Reverse the tuple and return it as a list
reversed_list = list(scores[::-1])
print("Reversed tuple as a list:", reversed_list)

# iv. Check if a score entered by the user is present
score = float(input("Enter a score to search: "))

if score in scores:
    print("Score found at index:", scores.index(score))
else:
    print("Score is not present in the tuple.")
```

```python
import random

# Create a list of 100 random numbers between 100 and 900
L = [random.randint(100, 900) for i in range(100)]

odd = []
even = []
prime = []

# Check each number
for n in L:
    if n % 2 != 0:
        odd.append(n)
    else:
        even.append(n)

    # Check whether the number is prime
    is_prime = True
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            is_prime = False
            break

    if is_prime:
        prime.append(n)

print("Random numbers:", L)

print("\nOdd numbers:", odd)
print("Count of odd numbers:", len(odd))

print("\nEven numbers:", even)
print("Count of even numbers:", len(even))

print("\nPrime numbers:", prime)
print("Count of prime numbers:", len(prime))
```

```python
A = {34, 56, 78, 90}
B = {78, 45, 90, 23}

# i. Union of sets
print("Union:", A.union(B))

# ii. Intersection of sets
print("Intersection:", A.intersection(B))

# iii. Symmetric difference
print("Symmetric Difference:", A.symmetric_difference(B))

# iv. Check subset and superset
print("Is A a subset of B?", A.issubset(B))
print("Is B a superset of A?", B.issuperset(A))

# v. Remove a specific score X from set A
X = int(input("Enter the score to remove from A: "))

if X in A:
    A.remove(X)
    print("Updated set A:", A)
else:
    print("Score is not present in set A.")
```
