```python

import pandas as pd

# Four fixed FAQ entries
fixed_entries = [
    {
        "question": "what is the annual fee",
        "answer": "The annual fee is Rs 500.",
        "keywords": "fee cost price charge",
        "category": "billing"
    },
    {
        "question": "how to reset password",
        "answer": "Go to Settings > Reset Password.",
        "keywords": "password reset login",
        "category": "account"
    },
    {
        "question": "what are your working hours",
        "answer": "We are open 9 AM to 5 PM.",
        "keywords": "hours timing open time",
        "category": "general"
    },
    {
        "question": "how can i pay the fee",
        "answer": "You can pay via UPI, card, or net banking.",
        "keywords": "pay payment upi fee",
        "category": "billing"
    }
]

# Input college roll number
roll_no = input("Enter your college roll number: ")

# Extract the last two digits
digits = [int(d) for d in roll_no[-2:]]

categories = ["billing", "account", "general"]

# Create two personalized FAQ entries
personalized_entries = []

for d in digits:
    category = categories[d % 3]

    if category == "billing":
        question = "how can I check my pending fee"
        answer = "You can check your pending fee in the billing section."
        keywords = "fee pending payment billing"

    elif category == "account":
        question = "how do I update my registered mobile number"
        answer = "Go to account settings to update your mobile number."
        keywords = "account mobile update number"

    else:
        question = "where can I find college office timings"
        answer = "College office timings are available on the official website."
        keywords = "college office timings general"

    personalized_entries.append({
        "question": question,
        "answer": answer,
        "keywords": keywords,
        "category": category
    })

# Combine fixed and personalized entries
all_entries = fixed_entries + personalized_entries

# Create DataFrame
df = pd.DataFrame(all_entries)

# Display the final 6-row DataFrame
print("\nPersonalized FAQ Knowledge Base:")
print(df)
```

```python

def score_query(query, df):
    query_words = set(query.lower().split())
    results = []

    for index, row in df.iterrows():
        keywords = set(row['keywords'].lower().split())
        question_words = set(row['question'].lower().split())

        # Count matching words in keywords and question
        matched_keywords = query_words & keywords
        matched_question = query_words & question_words

        # Calculate confidence score
        score = len(matched_keywords) + len(matched_question)

        # Store entries that match the query
        if score > 0:
            results.append({
                'question': row['question'],
                'answer': row['answer'],
                'category': row['category'],
                'score': score
            })

    # Sort by score in descending order
    results.sort(key=lambda x: x['score'], reverse=True)

    return results


# Test the function
query = input("Enter your query: ")

matches = score_query(query, df)

if matches:
    print("\nMatching FAQ entries:")
    for item in matches:
        print("\nQuestion:", item['question'])
        print("Answer:", item['answer'])
        print("Category:", item['category'])
        print("Confidence score:", item['score'])
else:
    print("No matching FAQ found.")
```
