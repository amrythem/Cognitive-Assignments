```python
roll_no = input("enter your roll no ")
L = [int(digit) * 10 for digit in roll_no]
print(L)
```

```python
L.append(21)
print(L)
```

```python
L.insert(5,69)
print(L)
```

```python
L.remove(40)
print(L)
```

```python
L.pop()
print(L)
```

```python
L.sort()
print(L)
```

```python
L.sort(reverse=True)
print(L)
```

```python
print(L[:3])
print(L[-3:])
```
